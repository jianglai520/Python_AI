```python
import pandas as pd

# 1.读取数据
data = pd.read_csv('data/sales.csv', nrows=10)
data
```

```python
# 排序
# 2.1 根据销售数量 倒序排序
data.sort_values(by='销售数量', ascending=False)

# 2.2 根据单价 升序排序
data.sort_values(by='单价', ascending=True)

# 2.3 根据价格 升序排序，价格一样，再根据 销售数量 倒序排序
data.sort_values(by=['单价', '销售数量'], ascending=[True, False])
```

```python
data = pd.read_csv("data/sales.csv", nrows = 20)

data['销售金额'] = data['单价'] * data['销售数量']
data
#3. 分组
# 3.1 根据 产品类别 分组，统计各个类别的 订单数量
data.groupby('产品类别')['订单号'].count()  #注意：缺失值不参与统计

# 3.2 根据 产品类别 分组，统计各个类别的 销售数量 之和
data.groupby('产品类别')['销售数量'].sum()

# 3.3 根据 产品类别 分组，统计各个类别的 销售金额 之和

data.groupby('产品类别')['销售金额'].sum()

# 3.4 根据 产品类别 分组，统计各个类别的最低商品 单价
data.groupby('产品类别')['单价'].min()

# 3.5 根据 产品类别 分组，统计各个类别的最高商品 单价
data.groupby('产品类别')['单价'].max()

# 3.6 根据 产品类别 分组，统计各个类别的平均商品 单价
data.groupby('产品类别')['单价'].mean()

# 3.7 根据 产品类别 分组，统计各个类别的商品的 平均单价、最高单价、最低单价
data.groupby('产品类别')['单价'].agg(['mean', 'max', 'min'])

# 3.8 根据 产品类别 分组，统计各个类别的商品的 销售数量 之和，销售金额 之和，平均 单价
data.groupby('产品类别').agg({'销售数量': 'sum', '销售金额': 'sum', '单价': 'mean'})
```
