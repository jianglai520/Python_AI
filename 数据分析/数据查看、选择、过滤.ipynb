```python
# 数据过滤
data = pd.read_csv('data/sales.csv', usecols = ['订单号', '产品类别', '产品名称', '销售数量', '单价'])
data

# 1.获取销售数量大于等于10 的订单数据
data[data['销售数量'] >= 10]

# 2.获取产品类别为食品 或者 图书的订单数据
data[data['产品类别'].isin(['食品', '图书'])]   #isin() 函数 --> 判断是否在列表中

# 3.单价在100-200之间的订单数据-->默认包含边界   # inclusive = True5
data[data['单价'].between(100, 200)]

# 4.获取销售数量 >= 8， 并且单价 >= 100 的订单数据
data[(data['销售数量'] >= 8) & (data['单价'] >= 100)]

# 5.获取产品类别为 服装/食品， 销售数量在 5~10 之间的订单数据
data[data['产品类别'].isin(['服装','食品']) & data['销售数量'].between(5, 10)]
```
