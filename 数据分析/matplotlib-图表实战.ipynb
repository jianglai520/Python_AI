```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt
from pydeck.data_utils.viewport_helpers import bbox_to_zoom_level

# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']  #设置默认字体为黑体

# 创建子图
# figure--画布对象, ax--子图数组
figure, ax = plt.subplots(nrows = 1, ncols = 2, figsize = (20, 6),dpi = 100)   # nrows--行数, ncols--列数, figsize--图片大小, dpi--图片分辨率

# 图一

# 数据准备
countries = ['美国', '中国', '日本', '印度', '法国', '英国', '俄罗斯', '德国', '意大利', '西班牙']

values = [2000000000, 1430000000, 1260000000, 1120000000, 1000000000, 900000000, 700000000, 700000000, 600000000, 500000000]

ax1: Axes = ax[0]
ax1.set_title('世界石油储备', fontsize = 20)  # fontsize--字体大小
ax1.bar(countries, values, width = 0.5, color = 'r')   # width--柱状宽度, color--柱状颜色
ax1.set_xlabel('国家')  # xlabel--X轴标签
ax1.set_ylabel('油量（亿桶）')  # ylabel--Y轴标签
ax1.grid(linestyle = '--', alpha = 0.3)  # linestyle--网格线样式, alpha--透明度

# 图二(饼状图)-- 世界人口
countries = ['美国', '中国', '日本', '印度', '法国', '其他', '俄罗斯']

values = [300000000, 1430000000, 1260000000, 1120000000, 1000000000, 3500000000, 700000000]

ax2: Axes = ax[1]
ax2.set_title('世界人口', fontsize = 20)
ax2.pie(values, labels = countries, autopct = '%1.1f%%', shadow = False)  # autopct--百分比格式, shadow--阴影
# ax2.axis('equal') # 设置饼状图等比例
ax2.legend(loc = 'lower center', ncol = 5, bbox_to_anchor = (0.5, -0.1))  # 图例 bbox_to_anchor--图例位置  ncol--每行显示的列数
# plt.subplots_adjust(wspace = 0.3) # 子图间距

# 保存图片
plt.savefig('data/01.png')


plt.show()  # 显示


```
