```python
import pandas as pd

pd.set_option('display.max_columns', 30)

# 1.读取数据
data = pd.read_csv('data/sales.csv')


# 2.数据清洗
# 2.1缺失值处理
data.isnull()   # 查看缺失值
# 2.1.1 删除缺失值
data.dropna()   # 删除缺失值所在行
data.dropna(axis=1)  # 删除缺失值所在列

# 2.1.2填充缺失值
data.fillna(0)  # 填充0
data.ffill()  # 填充上一行数据
data.bfill()  # 填充下一行数据
```

```python
#2.2 重复值处理
#2.2.1 查看重复值
data.duplicated()
data.duplicated(subset=['订单号'])
# #2.2.2 删除重复值
data.drop_duplicates(subset=['订单号'], keep = 'last')   # keep = 'last' 保留最后出现的重复值 keep = 'first' 保留首次出现的重复值
```

```python
# 2.3异常值处理
# 2.3.1 查看异常值
data[data['单价'] < 0]

# 2.3.2 删除异常值
data.drop(data[data['单价'] < 0].index)

# 2.3.3 修复异常值
data.loc[data['单价'] < 0, '单价'] = 29
data.loc[data['销售金额'] < 0, '销售金额'] = 203
data
```
