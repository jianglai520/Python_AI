```python
import matplotlib.pyplot as plt
import random


# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']  #设置默认字体为黑体


# 绘制折线图
x = [i for i in range(1, 25)]   # 注意：x轴的坐标值必须为数字，不能为字符串，否则会报错
y_bj = [random.randint(10, 15) for i in range(1, 25)] # x轴和y轴的数据量要一致
y_xi = [random.randint(5, 10) for i in range(1, 25)]

# 画布
plt.figure(figsize=(10, 6))   # 设置画布大小(宽，高)

# 设置折线图的详细信息

# 标题
plt.title("气温变化折线图", fontsize=15)  # fontsize设置标题的字体大小

# 轴标签
plt.xlabel("时间", fontsize=12)  # x轴标签
plt.ylabel("温度", fontsize=12)  # y轴标签

# 刻度标签
plt.xticks(x)
y_ticks = [i for i in range(0, 21)]
plt.yticks(y_ticks)

# 设置轴范围
plt.xlim(1, 24)
plt.ylim(0, 20)

# 绘制网格线
plt.grid( True, linestyle = '--', alpha = 0.3)




# 折线图
plt.plot(x,y_bj, label = "北京")
plt.plot(x,y_xi, label = "西宁")

#绘制图例
plt.legend(["北京", "西宁"], loc = "best")  # best表示自动选择位置,upper right表示右上角, upper left表示左上角, lower right表示右下角, lower left表示左下角, center right表示右中间, center left表示左中间, center表示中间, best表示自动选择位置

plt.show()  # 显示图表
```
